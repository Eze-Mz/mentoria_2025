## Kmeans + tfidf sustantivos

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
!wget -O df_final.csv "https://github.com/Eze-Mz/mentoria_2025/raw/main/datasets/df_final.csv"
```

```python
df = pd.read_csv("df_final.csv")

print(df.shape)
df.head()
```

```python
df.columns
```

Las variables que se eligen son:

1. sustantivos_lematizados: los sustantivos suelen concentrar el contenido temático de los tweets, hacen referencia a personas, instituciones, prácticas o fenómenos, palabras como salud, obesidad, sobrepeso. Al utilizar su forma lematizada se reduce la dispersión o variabilidad de las palabras, lo que facilita identificar agrupamientos en torno a temas recurrentes.

2. rawContent_clean: es el texto limpio del tuit, aunque trabajar con categorías gramaticales específicas ayuda a focalizar el análisis, mantener el texto completo normalizado es fundamental para técnicas como topic modeling, que capturan la co-ocurrencia de palabras en contexto.


Además de las representaciones basadas en frecuencias usando TF-IDF, se explorarán  Word2Vec y BERT embeddings. Se probarán estos tres enfoques diferentes y veremos ver las ventajas y desventajas de cada uno.

```python
import ast
df["sustantivos_lematizados_list"] = df["sustantivos_lematizados"].apply(ast.literal_eval)
df["sustantivos_lematizados_list"]
```

```python
df["sustantivos_text"] = df["sustantivos_lematizados_list"].apply(
    lambda lista: " ".join([palabra.replace(" ", "") for palabra in lista])
)
```

```python
df["sustantivos_text"]
```

```python
tfidf_sust = TfidfVectorizer(max_df=0.9, min_df=10)
X_sust = tfidf_sust.fit_transform(df["sustantivos_text"])
X_sust.shape
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []
K = range(2, 20)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_sust)
    inertia.append(kmeans.inertia_)


plt.figure(figsize=(6,4))
plt.plot(K, inertia, marker="o")
plt.xlabel("Número de clusters - k")
plt.ylabel("Inertia")
plt.title("Método del codo con sustantivos")
plt.show()
```

```python
from sklearn.metrics import silhouette_score


silhouettes = []

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_sust)
    score = silhouette_score(X_sust, kmeans.labels_)
    silhouettes.append(score)

plt.figure(figsize=(6,4))
plt.plot(K, silhouettes, marker='o', color="salmon")
plt.title("Silhouette score")
plt.xlabel("Número de clusters (k)")
plt.ylabel("Silhouette")

plt.show()
```

```python
k_sust = 5
kmeans_sust = KMeans(n_clusters=k_sust, random_state=42, n_init=10)
df["cluster_sust"] = kmeans_sust.fit_predict(X_sust)
```

```python

def top_words_per_cluster(tfidf_matrix, vectorizer, labels, n_words=10):
    df_words = pd.DataFrame(tfidf_matrix.todense(), columns=vectorizer.get_feature_names_out())
    df_words["cluster"] = labels
    top_words = {}
    for c in range(k_sust):
        cluster_mean = df_words[df_words["cluster"]==c].drop("cluster", axis=1).mean()
        top_words[c] = cluster_mean.sort_values(ascending=False).head(n_words).index.tolist()
    return top_words

top_words_sust = top_words_per_cluster(X_sust, tfidf_sust, df["cluster_sust"])
print("Sustantivos lematizados - top palabras por cluster:")
for c, words in top_words_sust.items():
    print(f"Cluster {c}: {words}")
```

```python
from sklearn.decomposition import TruncatedSVD
from sklearn.manifold import TSNE
import plotly.express as px
```

```python
from sklearn.decomposition import TruncatedSVD
import matplotlib.pyplot as plt
import numpy as np

svd = TruncatedSVD(n_components=100, random_state=42)
svd.fit(X_sust)

explained_variance = svd.explained_variance_ratio_
cum_explained_variance = np.cumsum(explained_variance)

plt.figure(figsize=(8,5))
plt.plot(range(1, len(cum_explained_variance)+1), cum_explained_variance, marker='o')
plt.axhline(y=0.90, color='r', linestyle='--', label='90% varianza')
plt.axhline(y=0.95, color='g', linestyle='--', label='95% varianza')
plt.xlabel('Número de componentes SVD')
plt.ylabel('Varianza explicada acumulada')
plt.title('Selección de número de componentes SVD')
plt.legend()
plt.grid(True)
plt.show()
```

El objetivo de aplicar TruncatedSVD es reducir la dimensionalidad de la matriz TF-IDF, que suele ser muy dispersa y de alta dimensión en el caso de texto.

Para decidir el número de componentes a conservar, se graficó la varianza explicada acumulada. Este criterio busca encontrar el punto en el que agregar más dimensiones ya no aporta una ganancia significativa en términos de información retenida.

En este caso, el gráfico mostró que incluso con 100 componentes sólo se capta alrededor del 50% de la varianza. Esto es esperable en datos de texto, ya que la información está distribuida en muchísimas dimensiones.

A pesar de ello, se seleccionó un número intermedio de 80 componentes como compromiso entre mantener la mayor información posible de los datos, reducir la dimensionalidad para que los algoritmos sean computacionalmente más manejables, evitar el exceso de componentes que podrían reintroducir ruido y dificultar la interpretación.

En resumen, la decisión de usar 80 componentes fue pragmática, si bien no se alcanza un umbral clásico entre 90 y 95% de varianza explicada, este valor permite conservar suficiente información para explorar patrones sin perder eficiencia computacional.

```python
svd_sust = TruncatedSVD(n_components=80, random_state=42)
X_sust_svd = svd_sust.fit_transform(X_sust)


tsne_sust = TSNE(n_components=2, perplexity=50, learning_rate=300, max_iter=2000, random_state=42) #1000, 30perplexity, learning rate 200,
X_sust_tsne = tsne_sust.fit_transform(X_sust_svd)


df_sust_plot = pd.DataFrame({
    "TSNE1": X_sust_tsne[:,0],
    "TSNE2": X_sust_tsne[:,1],
    "cluster": df["cluster_sust"].astype(str),
    "sustantivo": df["sustantivos_text"]
})


fig_sust = px.scatter(
    df_sust_plot, x="TSNE1", y="TSNE2", color="cluster",
    title="Clusters de Sustantivos lematizados - t-SNE",
    hover_data=["sustantivo" , "cluster"]
)
fig_sust.show()
```

Al analizar la métrica de inertia con el método del codo, se observó un punto de inflexión en k=5, lo que sugiere que este valor es un buen candidato inicial para el número de clusters. Sin embargo, al continuar incrementando k, la inertia mostró un comportamiento irregular, se puede observar que sube y baja, con una tendencia general a disminuir hasta k=20.

En cuanto al coeficiente de silhouette, el valor máximo alcanzado fue apenas 0.09 en k=20, lo cual es muy bajo y evidencia que los clusters formados con TF-IDF no están bien separados ni son consistentes. Incluso en k=5, el silhouette fue bajo, lo que refuerza la idea de que este enfoque no captura adecuadamente la estructura de los tuits.

Por lo tanto, aunque k=5 puede considerarse el número de clusters sugerido por el método del codo, los bajos valores de silhouette indican que la clusterización con TF-IDF no es efectiva en este corpus. Esto motiva avanzar hacia representaciones más ricas como Word2Vec o embeddings, que podrían mejorar la calidad de la agrupación.

Se continua con model_content5

Cada tweet se representa como el promedio de los vectores de las palabras que lo componen, obtenidos con Word2Vec. Esto permite transformar cada tweet en un vector único:

```python
import numpy as np

def tweet_embedding(tokens, model):
    vecs = []
    for token in tokens:
        if token in model.wv:
            vecs.append(model.wv[token])
    if len(vecs) == 0:
        return np.zeros(model.vector_size)
    return np.mean(vecs, axis=0)

tweet_embeddings = np.array([
    tweet_embedding(tweet.split(), model_content5)
    for tweet in df["rawContent_sin_simbolos"]
])
```

```python
k = 5
kmeans_w2v = KMeans(n_clusters=k, random_state=42, n_init=10)
labels = kmeans_w2v.fit_predict(tweet_embeddings)

df["w2v_cluster"] = labels
```

```python
tsne = TSNE(n_components=2, random_state=42, perplexity=40, max_iter=1000)
emb_2d = tsne.fit_transform(tweet_embeddings)

df_tsne = pd.DataFrame({
    "TSNE1": emb_2d[:, 0],
    "TSNE2": emb_2d[:, 1],
    "cluster": df["w2v_cluster"].astype(str),
    "rawContent": df["rawContent_clean"]})

fig = px.scatter(
    df_tsne,
    x="TSNE1",
    y="TSNE2",
    color="cluster",
    title="Clusters de Tweets completos - t-SNE",
    hover_data=["cluster", "rawContent"]
)

fig.show()
```

```python
def avg_cluster_similarity(embeddings, labels):
    """
    embeddings: array de vectores (tweets x dimensiones)
    labels: array con el cluster de cada tweet
    """
    clusters = np.unique(labels)
    avg_sims = {}

    for c in clusters:
        idx = np.where(labels == c)[0]
        cluster_emb = embeddings[idx]

        if len(cluster_emb) <= 1:
            avg_sims[c] = 1.0
            continue


        sim_matrix = cosine_similarity(cluster_emb)


        n = sim_matrix.shape[0]
        sum_sims = sim_matrix.sum() - n
        count = n * (n - 1)
        avg_sims[c] = sum_sims / count

    return avg_sims
```

#### sentence-transformers

La librería sentence-transformers es una capa de alto nivel construida sobre Hugging Face. Cuando cargamos un modelo de SBERT, la librería hace dos cosas:
- Descarga el modelo pre-entrenado desde el Hub de Hugging Face.
- Descarga el tokenizador correspondiente a ese modelo (ej. BertTokenizerFast).

**El Tokenizador es de Hugging Face**: El objeto `tokenizer_sbert` (en el código) es una instancia de una clase de la librería transformers, como AutoTokenizer. Usa el tokenizador oficial con el que el modelo fue entrenado.

Cuando ejecutamos `tokenizer_sbert(docs, ...)`, ocurren los siguientes pasos para cada texto en la lista docs:
- Tokenización: El texto se divide en "tokens" (palabras o sub-palabras) según las reglas del modelo (ej. WordPiece para BERT). "Hola mundo" -> ['hola', 'mun', '##do'].
- Conversión a IDs: Cada token se mapea a un número entero (ID) del vocabulario del modelo. ['hola', 'mun', '##do'] -> [8324, 1525, 2255].
- Añadir Tokens Especiales: Si add_special_tokens=True, se añaden los tokens que el modelo necesita para entender la estructura de la secuencia, como [CLS] al inicio y [SEP] al final.
- Creación del Batch: El resultado de todos los textos se agrupa en un objeto BatchEncoding, que es básicamente un diccionario que contiene listas de input_ids, attention_mask (si se solicita), etc.

```python
def embed_query(query: str, model, tokenizer, device):
    """
    Genera el embedding para una única consulta de texto.
    """
    # El tokenizador espera una lista, así que pasamos la consulta entre corchetes
    inputs = tokenizer(
        [query],
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=128
    )

    # Mueve los inputs al dispositivo correcto (CPU o GPU)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    # Obtiene la salida del modelo sin calcular gradientes
    with torch.no_grad():
        outputs = model(**inputs)

    # Extrae el embedding del token [CLS], lo mueve a la CPU y lo convierte a NumPy
    embedding = outputs.last_hidden_state[:, 0, :].cpu().numpy()

    return embedding
```

```python
import re

def tokenizar_para_fasttext(texto):
    """
    Tokeniza el texto de una forma más similar a como lo procesa FastText:
    1. Convierte a minúsculas.
    2. Separa palabras de signos de puntuación.
    """
    texto = texto.lower()
    # Esta expresión regular encuentra secuencias de palabras o signos de puntuación individuales
    tokens = re.findall(r'\b\w+\b|[.,!?;:()]', texto)
    return tokens

def generar_embeddings_documentos(lista_de_documentos, modelo):
    """
    Genera embeddings para una lista de documentos usando average pooling.
    """
    embeddings = []
    dim = modelo.get_dimension() # Obtenemos la dimensión de los vectores del modelo

    for doc in lista_de_documentos:
        tokens_split = doc.split()
        tokens_regex = tokenizar_para_fasttext(doc)
        #print(f"Tokens (split): {tokens_split}")
        #print(f"Tokens (regex): {tokens_regex}")
        tokens = tokens_regex  # Elegimos la tokenización basada en regex para mayor precisión
        if not tokens:
            # Si un documento está vacío, agregamos un vector de ceros
            embeddings.append(np.zeros(dim))
            continue

        # Obtenemos el vector de cada token y calculamos el promedio
        # np.mean es muy eficiente para esto
        vector_promedio = np.mean([modelo.get_word_vector(token) for token in tokens], axis=0)
        embeddings.append(vector_promedio)

    return np.array(embeddings)
```

```python
from umap import UMAP
from hdbscan import HDBSCAN
from sklearn.feature_extraction.text import CountVectorizer
from bertopic.vectorizers import ClassTfidfTransformer
import nltk
from nltk.corpus import stopwords
from bertopic.representation import KeyBERTInspired, MaximalMarginalRelevance
from bertopic import BERTopic
```

```python
def calculate_topics(text, embeddings, umap_neighbors=10, umap_components=6, hdbscan_size=5, embedding_model=None):
  # https://umap-learn.readthedocs.io/en/latest/parameters.html
  # low values of n_neighbors will force UMAP to concentrate on very local structure (potentially to the detriment of the big picture)
  # low values of min_dist will result in clumpier embeddings. This can be useful if you are interested in clustering, or in finer topological structure.
  # If you are interested in (density based) clustering, or other machine learning techniques, it can be beneficial to pick a larger embedding dimension (say 10, or 50)
  # metrics: euclidean, cosine, manhattan, chebishev, minkowski, etc.
  umap_model = UMAP(n_neighbors=umap_neighbors, n_components=umap_components, min_dist=0.0, metric='cosine', random_state=42)

  # https://hdbscan.readthedocs.io/en/latest/parameter_selection.html
  # The default method is 'eom' for Excess of Mass, has a tendency to pick one or two large clusters and then a number of small extra clusters.
  # Instead, a better option is to select 'leaf' as a cluster selection method. This will select leaf nodes from the tree.
  hdbscan_model = HDBSCAN(min_cluster_size=hdbscan_size,
                          metric='euclidean',
                          cluster_selection_method='leaf',
                          leaf_size=100,
                          min_samples=1,
                          prediction_data=True)

  spanish_stopwords = set(stopwords.words('spanish'))
  custom_stopwords = {'ejemplo','ser','estar','así','solo','vez','mientras','ud','si','aquí','eh', 'bueno', 'va', 'ahí','vamos','bien','ver',
                      'entonces','día','hoy','voy','acá','vol'}
  spanish_stopwords.update(custom_stopwords)
  spanish_stopwords = list(spanish_stopwords)

  vectorizer_model = CountVectorizer(stop_words=spanish_stopwords, min_df=0.5, max_df=0.8, ngram_range=(1, 2))

  ctfidf_model = ClassTfidfTransformer(
      bm25_weighting=False,
      reduce_frequent_words=False
  )


  # Agregar diferentes representaciones
  # KeyBERT
  keybert_model = KeyBERTInspired()
  # MMR
  mmr_model = MaximalMarginalRelevance(diversity=0.5)

  representation_model = {
      "KeyBERT": keybert_model,
      "MMR": mmr_model,
  }

  topic_model = BERTopic(
  # Modelos del pipeline
  embedding_model=embedding_model,
  umap_model=umap_model,
  hdbscan_model=hdbscan_model,
  vectorizer_model=vectorizer_model,
  representation_model=representation_model,
  ctfidf_model=ctfidf_model,
  # Otras opciones
  top_n_words=10,
  verbose=True,
  calculate_probabilities=True
)

  topics, probs = topic_model.fit_transform(text, embeddings)
  return topics, probs, topic_model
```

```python
def inspect_topic(docs_df, topics_df, topic_id, n_documents=5):
    """
    Imprime un resumen de un tópico específico, incluyendo sus representaciones
    y los documentos más representativos.

    Args:
        df (pd.DataFrame): DataFrame con documentos, tópicos y probabilidades.
        topic_id (int): El ID del tópico que se quiere inspeccionar.
        n_documents (int): El número de documentos a mostrar.
    """
    # Filtrar el DataFrame para obtener solo los datos del tópico seleccionado
    topic_docs_df = docs_df[docs_df['topic'] == topic_id].copy()

    if topic_docs_df.empty:
        print(f"ERROR: Tópico {topic_id} no encontrado en el DataFrame.")
        return

    # Ordenar los documentos por la probabilidad del tópico en orden descendente
    top_docs = topic_docs_df.sort_values(by='topic_prob', ascending=False).head(n_documents)

    # Extraer las representaciones (son las mismas para todo el tópico)
    topic_info_df = topics_df[topics_df['Topic'] == topic_id].copy().iloc[0]

    # --- Imprimir la información ---
    print(f"======= INFORME DEL TÓPICO: {topic_id} =======")
    print(f"Total de documentos en el tópico: {len(topic_docs_df)}\n")

    print("--- REPRESENTACIONES DE PALABRAS CLAVE ---")
    print(f"Principal    : {topic_info_df['Representation']}")
    print(f"KeyBERT      : {topic_info_df['KeyBERT']}")
    print(f"MMR          : {topic_info_df['MMR']}\n")

    print(f"--- {n_documents} DOCUMENTOS MÁS REPRESENTATIVOS ---")
    for index, row in top_docs.iterrows():
        prob = row['topic_prob']
        doc = row['document']
        print(f"Prob: {prob:.4f} -> {doc}")

    print("=" * (30 + len(str(topic_id))))
```

```python
import circlify
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import textwrap

def plot_packed_bubbles_circlify(df, selected_topics, use_title=False, figsize=(8, 8)):
    """
    Gráfico de burbujas empaquetadas usando circlify con text-wrap y mejor distribución
    """
    # filter with selected topics
    df = df[df['Topic'].isin(selected_topics)]
    counts = df['Count']

    # Preparar datos para circlify
    data = []
    for topic, count in counts.items():
        data.append({
            'id': topic,
            'title': df_to_plot.loc[topic, 'custom_labels'],
            'datum': count,
        })

    # Calcular círculos empaquetados con factor de escala mayor
    scale_factor = 1.3  # Aumentamos el tamaño
    data_cleaned = [{'id': item['id'], 'datum': item['datum']} for item in data]
    # OJO: ordena de menor a mayor radio
    circles = circlify.circlify(data_cleaned, show_enclosure=False)

    # Colores
    pastel_colors = [
        "#FFB3BA", "#FFDFBA", "#FFFFBA", "#BAFFC9", "#BAE1FF",
        "#E1BBFF", "#FFD1DC", "#C2F0C2", "#F7D6E0", "#D0F0F7",
        "#FBE7C6", "#E8D9FF", "#C9E7FF", "#FFE4E1", "#E6FFE6",
        "#FFF0B3", "#D6EAF8", "#FADBD8", "#D5F5E3", "#FDEBD0"
    ]

    # Crear un diccionario que asigne un color a cada topic (repite si hay >20)
    pal = {}
    topics = list(counts.index)
    for i, topic in enumerate(topics):
        pal[topic] = pastel_colors[i % len(pastel_colors)]

    # Crear gráfico con límites más amplios
    fig, ax = plt.subplots(figsize=figsize)

    def wrap_text(text, max_chars_per_line):
        """Función auxiliar para hacer text wrap"""
        return '\n'.join(textwrap.wrap(text, width=max_chars_per_line, break_long_words=False))

    # Crear un diccionario para buscar datos por ID fácilmente
    data_map = {item['id']: item for item in data}

    # Dibujar círculos
    for circle in circles:
        # Obtener el ID del círculo actual
        item_id = circle.ex['id']
        # Usar el ID para obtener la información correcta del diccionario
        item = data_map[item_id]

        x, y, r = circle.x, circle.y, circle.r
        color = pal[item_id]

        circle_patch = patches.Circle((x, y), r, facecolor=color,
                                    edgecolor='white', linewidth=3, alpha=0.8)
        ax.add_patch(circle_patch)

        top_y = y + r * 0.35
        middle_y = y - r * 0.35

        title_fontsize = max(8, min(12, r * 60))
        number_fontsize = max(10, min(16, r * 80))
        chars_per_line_title = max(10, int(r * 80))

        title_wrapped = wrap_text(item['title'], chars_per_line_title)
        ax.text(x, top_y, title_wrapped, ha='center', va='center',
                fontsize=title_fontsize, weight='bold', color='white',
                bbox=dict(boxstyle="round,pad=0.3", facecolor='black', alpha=0.7))

        ax.text(x, middle_y, f"{item['datum']}", ha='center', va='center',
                fontsize=number_fontsize, weight='bold', color='white',
                bbox=dict(boxstyle="round,pad=0.2", facecolor='black', alpha=0.5))

    # Configurar ejes con límites más amplios para acomodar círculos más grandes
    xs = [c.x - c.r for c in circles] + [c.x + c.r for c in circles]
    ys = [c.y - c.r for c in circles] + [c.y + c.r for c in circles]
    x_min, x_max = min(xs), max(xs)
    y_min, y_max = min(ys), max(ys)
    # añadir un pequeño margen
    margin = 0.05 * max(x_max - x_min, y_max - y_min)
    ax.set_xlim(x_min - margin, x_max + margin)
    ax.set_ylim(y_min - margin, y_max + margin)
    ax.set_aspect('equal')
    ax.axis('off')

    if use_title:
        ax.set_title('Distribución de Menciones por Categoría',
                     fontsize=18, weight='bold', pad=20)

    plt.tight_layout()
    return fig, ax
```

```python
import matplotlib.pyplot as plt

def plot_topic_sentiment(df, topic_number, figsize=(4, 4), palette=None, title=None):
    """
    Grafica el sentimiento (pysentimiento) de un tópico como barras verticales.
    - df: DataFrame que contiene las columnas 'topic' y 'pysentimiento'
    - topic_number: número del tópico a graficar
    - figsize: tamaño de la figura (por defecto pequeño)
    - palette: dict opcional {'positive': color, 'neutral': color, 'negative': color}
    - title: título opcional
    Retorna: fig, ax
    """
    # Palette por defecto (pastel)
    default_palette = {
        'positive': "#BAFFC9",  # verde pastel
        'neutral' : "#FFFFBA",  # amarillo pastel
        'negative': "#FFB3BA",  # rojo pastel
    }
    if palette is None:
        palette = default_palette
    else:
        # llenar con defaults si falta alguna clave
        for k, v in default_palette.items():
            palette.setdefault(k, v)

    # Filtrar por tópico
    sub = df[df['topic'] == topic_number]
    if sub.shape[0] == 0:
        raise ValueError(f"No se encontraron filas para el tópico {topic_number}")

    # Normalizar etiquetas de pysentimiento a: 'positive', 'neutral', 'negative'
    def _norm_sent(x):
        if x is None:
            return 'neutral'
        s = str(x).strip().lower()
        # checks simples y robustos
        if s.startswith('pos') or 'positivo' in s or 'positive' in s or s == 'p' or s == '+':
            return 'positive'
        if s.startswith('neg') or 'negativo' in s or 'negative' in s or s == 'n' or s == '-':
            return 'negative'
        if s.startswith('neu') or 'neutral' in s or s == '0':
            return 'neutral'
        # fallback: buscar palabras clave
        if 'positivo' in s or 'positive' in s:
            return 'positive'
        if 'negativo' in s or 'negative' in s:
            return 'negative'
        return 'neutral'

    sentiments = sub['pysentimiento'].apply(_norm_sent)

    # Contar y reindexar en orden: negative, neutral, positive (para color consistente)
    order = ['negative', 'neutral', 'positive']
    counts = sentiments.value_counts().reindex(order).fillna(0).astype(int)

    # Plot
    fig, ax = plt.subplots(figsize=figsize)
    labels_es = {'negative': 'Negativo', 'neutral': 'Neutro', 'positive': 'Positivo'}
    colors = [palette[k] for k in counts.index]
    bars = ax.bar([labels_es[k] for k in counts.index], counts.values, color=colors)

    # Título
    total = counts.sum()
    ax.set_title(title or f"Sentimiento - tópico {topic_number} (n={total})", fontsize=10, weight='bold')
    ax.set_ylabel("Cantidad", fontsize=9)

    # Anotaciones: conteo y porcentaje
    for bar, cnt in zip(bars, counts.values):
        pct = (cnt / total * 100) if total > 0 else 0
        ax.annotate(f"{cnt}\n{pct:.1f}%",
                    xy=(bar.get_x() + bar.get_width() / 2, cnt),
                    xytext=(0, 3), textcoords="offset points",
                    ha="center", va="bottom", fontsize=8)

    # Estética pequeña
    ax.set_ylim(0, max(1, counts.max() * 1.15))
    ax.grid(axis='y', linestyle='--', alpha=0.25)
    plt.tight_layout()
    return fig, ax
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
from matplotlib.transforms import Affine2D
import textwrap
import random
import pandas as pd
import numpy as np
import re

def plot_topic_cards_simple(docs_df, topic_id, indices=None, max_cards=5,
                            figsize=(14, 7), color=None, seed=42, fontsize=11):
    """
    Versión mejorada: muestra hasta `max_cards` tarjetas del tópico `topic_id`.
    - indices: posiciones en el ranking por topic_prob (0 = top1). Si None toma [0,1,2].
    - color: string hex (ej. '#BAFFC9') para usar el mismo color en todas las tarjetas.
    - fontsize: tamaño base de la fuente (default=11)

    Layout: 2 tarjetas arriba, 3 abajo (o menos según max_cards)
    Cada tarjeta ajusta su altura según el contenido del texto.
    """
    # Color mejorado - más parecido a post-its
    default_color = "#FFF9C4"  # Amarillo suave tipo post-it
    face_color = color if color is not None else default_color
    edge_color = "#424242"  # Gris oscuro para el borde

    # Filtrar tópico
    topic_docs = docs_df[docs_df['topic'] == topic_id].copy()
    if topic_docs.empty:
        raise ValueError(f"No se encontraron documentos para el tópico {topic_id}")

    # Ordenar por probabilidad (desc)
    if 'topic_prob' in topic_docs.columns:
        topic_docs = topic_docs.sort_values(by='topic_prob', ascending=False).reset_index(drop=True)
    else:
        topic_docs = topic_docs.reset_index(drop=True)

    # Indices por defecto si no proporcionados
    if indices is None:
        indices = list(range(min(3, len(topic_docs))))
    # Filtrar índices válidos y limitar a max_cards
    valid_indices = [i for i in indices if isinstance(i, int) and 0 <= i < len(topic_docs)]
    if not valid_indices:
        raise ValueError("La lista de índices no contiene posiciones válidas dentro del tópico.")
    valid_indices = valid_indices[:max_cards]

    selected = topic_docs.loc[valid_indices].reset_index(drop=True)

    # Preparar figura con fondo blanco
    fig, ax = plt.subplots(figsize=figsize, facecolor='white')
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis('off')
    ax.set_facecolor('white')

    n = len(selected)
    rand = random.Random(seed)

    # Helper: limpiar caracteres raros y control chars
    def clean_text(s):
        if pd.isna(s) or s is None:
            return ""
        t = str(s)
        t = t.replace('\ufeff', '').replace('\ufffd', '')
        t = re.sub(r'[\x00-\x1f\x7f-\x9f]', '', t)
        t = t.strip()
        return t

    # Determinar layout: 2 arriba, resto abajo
    n_top = min(2, n)
    n_bottom = n - n_top

    # Anchos de tarjetas - fijos para evitar superposición
    card_w_top = 0.35
    card_w_bottom = 0.24

    # Posiciones base X e Y para cada fila
    y_top = 0.66
    y_bottom = 0.45

    # Márgenes
    margin_h = 0.10

    positions_base = []
    card_widths = []

    # Fila superior (2 tarjetas)
    if n_top > 0:
        if n_top == 1:
            x = 0.5 - card_w_top / 2
            positions_base.append((x, y_top))
            card_widths.append(card_w_top)
        else:
            spacing = (1 - 2 * margin_h - 2 * card_w_top) / 1
            x1 = margin_h
            x2 = margin_h + card_w_top + spacing
            positions_base.append((x1, y_top))
            positions_base.append((x2, y_top))
            card_widths.extend([card_w_top, card_w_top])

    # Fila inferior (hasta 3 tarjetas)
    if n_bottom > 0:
        if n_bottom == 1:
            x = 0.5 - card_w_bottom / 2
            positions_base.append((x, y_bottom))
            card_widths.append(card_w_bottom)
        elif n_bottom == 2:
            spacing = (1 - 2 * margin_h - 2 * card_w_bottom) / 1
            x1 = margin_h + (card_w_top - card_w_bottom) / 2
            x2 = x1 + card_w_bottom + spacing
            positions_base.append((x1, y_bottom))
            positions_base.append((x2, y_bottom))
            card_widths.extend([card_w_bottom, card_w_bottom])
        else:
            spacing = (1 - 2 * margin_h - 3 * card_w_bottom) / 2
            x1 = margin_h
            x2 = x1 + card_w_bottom + spacing
            x3 = x2 + card_w_bottom + spacing
            positions_base.append((x1, y_bottom))
            positions_base.append((x2, y_bottom))
            positions_base.append((x3, y_bottom))
            card_widths.extend([card_w_bottom, card_w_bottom, card_w_bottom])

    # Calcular contenido y ALTURA de cada tarjeta según el texto
    card_dims = []
    for i, orig_idx in enumerate(valid_indices):
        row = selected.iloc[i]
        card_w = card_widths[i]

        # Obtener texto del documento
        doc_text = ""
        for col in ['document', 'rawContent', 'rawContent_clean', 'rawContent_clean_lower']:
            if col in row and pd.notna(row[col]):
                doc_text = clean_text(row[col])
                break
        if not doc_text:
            doc_text = "<sin texto disponible>"

        # Determinar ancho en caracteres según card_w
        chars_per_line = max(30, int(card_w * 110))
        wrapped = textwrap.fill(doc_text, width=chars_per_line,
                                break_long_words=False, break_on_hyphens=False)

        # CALCULAR ALTURA según número de líneas
        n_lines = wrapped.count('\n') + 1
        line_height = 0.022  # Altura por línea de texto
        padding_vertical = 0.06  # Padding superior + inferior

        # Altura dinámica con límites razonables
        card_h = min(0.35, max(0.14, n_lines * line_height + padding_vertical*1.2))

        card_dims.append({
            'width': card_w,
            'height': card_h,
            'text': wrapped,
            'orig_idx': orig_idx
        })

    # Posiciones finales (sin jitter para evitar superposición)
    positions = []
    for i, (x_base, y_base) in enumerate(positions_base):
        card_w = card_dims[i]['width']
        card_h = card_dims[i]['height']

        # Usar posiciones exactas sin jitter
        x = x_base
        y = y_base

        # Asegurar que está dentro de límites
        x = np.clip(x, 0.02, 1 - card_w - 0.02)
        y = np.clip(y, 0.05, 0.95 - card_h)

        positions.append((x, y, card_w, card_h))

    # Sin rotación para mantener alineamiento perfecto
    rotations = [0 for _ in range(n)]

    # Dibujar cada tarjeta
    for i in range(n):
        card_data = card_dims[i]
        left, bottom, card_w, card_h = positions[i]

        # Sombra sutil
        shadow = FancyBboxPatch((left + 0.008, bottom - 0.008), card_w, card_h,
                                boxstyle="round,pad=0.01,rounding_size=0.02",
                                transform=ax.transAxes,
                                linewidth=0, facecolor='#000000',
                                alpha=0.15, zorder=1)
        t_shadow = Affine2D().rotate_deg_around(left + card_w/2 + 0.008,
                                                  bottom + card_h/2 - 0.008,
                                                  rotations[i]) + ax.transAxes
        shadow.set_transform(t_shadow)
        ax.add_patch(shadow)

        # Caja redondeada principal
        bbox = FancyBboxPatch((left, bottom), card_w, card_h,
                              boxstyle="round,pad=0.01,rounding_size=0.02",
                              transform=ax.transAxes,
                              linewidth=2, facecolor=face_color,
                              edgecolor=edge_color, alpha=0.98,
                              zorder=3)

        # Sin rotación
        t = Affine2D().rotate_deg_around(left + card_w/2, bottom + card_h/2, rotations[i]) + ax.transAxes
        bbox.set_transform(t)
        ax.add_patch(bbox)

        # Padding interno
        pad_x = 0.025
        pad_y_top = 0.03

        # Texto principal
        text_x = left + pad_x
        text_y = bottom + card_h - pad_y_top

        ax.text(text_x, text_y, card_data['text'],
                transform=t, fontsize=fontsize, va='top', ha='left',
                color='#212121', linespacing=1.2, zorder=5,
                family='sans-serif')

    plt.tight_layout(pad=0.5)
    return fig, ax

fig, ax = plot_topic_cards_simple(df_with_topics, topic_id=4, indices=[0,1,2,3,4], fontsize=11)
plt.show()
```

```python
# Requiere: pip install wordcloud
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt
import re
import pandas as pd
import numpy as np

def topic_wordcloud(docs_df, topic_id, n_docs=50, max_words=100,
                    palette=None, figsize=(10,6), background_color='white',
                    extra_stopwords=None, seed=42, collocations=False):
    """
    Genera una nube de palabras con los primeros `n_docs` documentos del tópico `topic_id`.

    Args:
        docs_df: DataFrame con al menos columnas ['topic', 'topic_prob' (opcional), 'document'|'rawContent'...].
        topic_id: int, id del tópico a filtrar.
        n_docs: int, cantidad de documentos a tomar del ranking (por defecto 50).
        max_words: int, número máximo de palabras a mostrar en la nube.
        palette: list de hex colors pastel (ej. ['#FFB3BA', ...]). Si None usa paleta por defecto.
        figsize: tamaño de la figura matplotlib.
        background_color: fondo de la imagen.
        extra_stopwords: iterable con stopwords extra (opcional).
        seed: semilla para reproducibilidad.
        collocations: si True, WordCloud agrupa collocations (por defecto False).
    Returns:
        fig, ax (matplotlib)
    """
    # --- paleta pastel por defecto (20)
    pastel_default = [
    "#A8D5BA",  # verde menta apagado
    "#F5C6AA",  # durazno suave
    "#C4B7CB",  # lavanda grisácea
    "#FFD5A5",  # crema cálida
    "#B5D8EB",  # celeste apagado
    "#E2B6CF",  # rosa viejo
    "#D0CFCF",  # gris cálido
    "#AFC8AD",  # verde seco
    ]
    colors = pastel_default if palette is None else list(palette)

    # Filtrar tópico
    topic_docs = docs_df[docs_df['topic'] == topic_id].copy()
    if topic_docs.empty:
        raise ValueError(f"No se encontraron documentos para el tópico {topic_id}")

    # Ordenar por topic_prob si existe
    if 'topic_prob' in topic_docs.columns:
        topic_docs = topic_docs.sort_values(by='topic_prob', ascending=False).reset_index(drop=True)
    else:
        topic_docs = topic_docs.reset_index(drop=True)

    # Tomar hasta n_docs
    top_docs = topic_docs.head(n_docs)

    # Extraer texto (preferir columnas conocidas)
    text_pieces = []
    for _, row in top_docs.iterrows():
        text = None
        for col in ['document', 'rawContent', 'rawContent_clean', 'rawContent_clean_lower']:
            if col in row and pd.notna(row[col]):
                text = str(row[col])
                break
        if text is None:
            text = ''
        # limpieza básica por documento
        # quitar URLs, menciones, hashtags sueltos, y caracteres de control
        text = re.sub(r'http\S+|https?\S+', ' ', text)
        text = re.sub(r'@\w+', ' ', text)
        text = re.sub(r'#', ' ', text)  # quitar solo el # dejando la palabra
        text = re.sub(r'[\x00-\x1f\x7f-\x9f]', ' ', text)
        text = text.strip()
        if text:
            text_pieces.append(text)

    if not text_pieces:
        raise ValueError("No hay texto válido en los documentos seleccionados.")

    combined_text = " ".join(text_pieces).lower()

    # Stopwords: union de wordcloud STOPWORDS + extra_stopwords
    stopwords = set(STOPWORDS)
    if extra_stopwords:
        stopwords.update([w.lower() for w in extra_stopwords])

    # WordCloud
    wc = WordCloud(width=1200, height=700,
                   background_color=background_color,
                   stopwords=stopwords,
                   max_words=max_words,
                   collocations=collocations,
                   random_state=seed).generate(combined_text)

    # Color function: asigna color pastel basado en hash de la palabra (determinístico)
    def pastel_color_func(word, font_size, position, orientation, random_state=None, **kwargs):
        idx = abs(hash(word)) % len(colors)
        return colors[idx]

    # Plot
    fig, ax = plt.subplots(figsize=figsize)
    ax.imshow(wc.recolor(color_func=pastel_color_func, random_state=seed), interpolation='bilinear')
    ax.axis('off')
    plt.tight_layout()
    return fig, ax

# cargar las stopwords en español de ntlk
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
stopwords_es = stopwords.words('spanish')
stopwords_es.extend({'obeso','obesa','sobrepeso','obesidad','gordo','gorda','pasa','vos','puede','obesos', 'q', 'día', 'solo', 'si', 'chau', 'pene', 'hombre'})

fig, ax = topic_wordcloud(df_with_topics, topic_id=13, n_docs=100, max_words=30, extra_stopwords=stopwords_es)
plt.show()
```
