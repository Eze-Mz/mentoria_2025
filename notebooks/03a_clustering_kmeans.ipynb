```python
clustering_metrics = []
clustering_assignments = []
clustering_topic_terms = []

def centroid_assignment_score(model, X):
    """Devuelve un score simple basado en distancia al centroide: más alto implica mayor cercanía."""
    distances = model.transform(X)
    min_distance = distances.min(axis=1)
    return 1 / (1 + min_distance)

def register_clustering_result(method, variant, labels, representation, assignment_score=None, notes=None):
    labels = np.asarray(labels)
    sizes = pd.Series(labels).value_counts().sort_index()
    row = {
        "method": method,
        "variant": variant,
        "n_clusters": int(len(np.unique(labels))),
        "min_cluster_size": int(sizes.min()),
        "median_cluster_size": float(sizes.median()),
        "max_cluster_size": int(sizes.max()),
        "silhouette": float(silhouette_score(representation, labels)),
        "davies_bouldin": float(davies_bouldin_score(representation, labels)),
        "calinski_harabasz": float(calinski_harabasz_score(representation, labels)),
        "notes": notes or "",
    }
    clustering_metrics.append(row)

    assignments = pd.DataFrame({
        "doc_id": df.index,
        "method": method,
        "variant": variant,
        "topic": labels,
    })
    if assignment_score is not None:
        assignments["assignment_score"] = assignment_score
    clustering_assignments.append(assignments)
    return row



TOPIC_DESCRIPTOR_STOPWORDS = {
    "obesidad", "obeso", "obesa", "obesos", "obesas",
    "sobrepeso", "gordo", "gorda", "gordos", "gordas", "gordura",
    "mórbida", "morbida", "mórbido", "morbido",
    "ser", "tener", "hacer", "hace", "va", "re", "mas", "más", "solo",
    "gente", "persona", "personas", "cosa", "cosas", "día", "dias", "días",
    "año", "años", "vez", "bien", "mal", "puede", "ahora",
}

def top_terms_from_cluster_documents(texts, labels, n_terms=12, stop_words=None, ngram_range=(1, 2)):
    """
    Calcula términos representativos a nivel cluster.

    En lugar de promediar TF-IDF de documentos individuales, concatena los textos
    de cada cluster y calcula TF-IDF entre clusters. Esto funciona como una versión
    simple de c-TF-IDF: penaliza términos que aparecen en muchos clusters y sube
    los que distinguen a cada grupo.
    """
    cluster_docs = pd.DataFrame({
        "text": pd.Series(texts).astype(str),
        "topic": np.asarray(labels),
    })
    cluster_docs = (
        cluster_docs
        .groupby("topic", as_index=False)["text"]
        .agg(" ".join)
        .sort_values("topic")
        .reset_index(drop=True)
    )

    vectorizer = TfidfVectorizer(
        stop_words=list(stop_words) if stop_words is not None else None,
        ngram_range=ngram_range,
        min_df=1,
        max_df=0.95,
        sublinear_tf=True,
    )
    matrix = vectorizer.fit_transform(cluster_docs["text"])
    feature_names = np.array(vectorizer.get_feature_names_out())

    rows = []
    for row_idx, topic in enumerate(cluster_docs["topic"]):
        weights = np.asarray(matrix[row_idx].todense()).ravel()
        top_indices = weights.argsort()[::-1][:n_terms]
        for rank, idx in enumerate(top_indices, start=1):
            rows.append({
                "topic": int(topic),
                "rank": rank,
                "term": feature_names[idx],
                "weight": float(weights[idx]),
            })
    return pd.DataFrame(rows)

def register_cluster_terms(method, variant, terms_df):
    terms_df = terms_df.copy()
    terms_df.insert(0, "method", method)
    terms_df.insert(1, "variant", variant)
    clustering_topic_terms.append(terms_df)
    return terms_df
```

KMeans busca separar los documentos en grupos minimizando la suma de distancias cuadráticas entre cada observación y el centroide de su cluster:

$$\min_{C_1,\dots,C_k} \sum_{j=1}^{k} \sum_{x_i \in C_j} ||x_i - \mu_j||^2$$

Por eso la calidad del resultado depende mucho de cómo se represente cada tweet: si la representación no captura contexto, KMeans solo va a separar por coincidencias superficiales de palabras.


TF-IDF pondera cada término combinando su frecuencia en un documento con su rareza en el corpus:

$$tfidf(t,d)=tf(t,d)\times \log\left(\frac{N}{df(t)}\right)$$

Esta representación es útil como línea base porque es simple e interpretable, aunque suele perder significado contextual en textos cortos y ruidosos como tweets.


Las variables que se eligen son:

Al utilizar su forma lematizada se reduce la dispersión o variabilidad de las palabras, lo que facilita identificar agrupamientos en torno a temas recurrentes.

2. rawContent_clean: es el texto limpio del tuit, aunque trabajar con categorías gramaticales específicas ayuda a focalizar el análisis, mantener el texto completo normalizado es fundamental para técnicas como topic modeling, que capturan la co-ocurrencia de palabras en contexto.

```python
tfidf_sust = TfidfVectorizer(max_df=0.9, min_df=10)
X_sust = tfidf_sust.fit_transform(df["sustantivos_text"])
X_sust.shape
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

K = range(2, 61)
inertia = []

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_sust)
    inertia.append(kmeans.inertia_)

k_values = np.array(list(K))
inertia_values = np.array(inertia)
relative_inertia_drop = np.r_[np.nan, (inertia_values[:-1] - inertia_values[1:]) / inertia_values[:-1]]

# Heurística simple de codo: máxima distancia a la recta entre el primer y último punto normalizados.
x_norm = (k_values - k_values.min()) / (k_values.max() - k_values.min())
y_norm = (inertia_values - inertia_values.min()) / (inertia_values.max() - inertia_values.min())
line_start = np.array([x_norm[0], y_norm[0]])
line_end = np.array([x_norm[-1], y_norm[-1]])
points = np.column_stack([x_norm, y_norm])
line_vector = line_end - line_start
rel_points = line_start - points
elbow_distances = np.abs(line_vector[0] * rel_points[:, 1] - line_vector[1] * rel_points[:, 0]) / np.linalg.norm(line_vector)
k_elbow_sust = int(k_values[np.argmax(elbow_distances)])

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].plot(k_values, inertia_values, marker="o")
axes[0].axvline(k_elbow_sust, color="crimson", linestyle="--", label=f"codo aprox.: k={k_elbow_sust}")
axes[0].set_xlabel("Número de clusters - k")
axes[0].set_ylabel("Inertia")
axes[0].set_title("Método del codo con sustantivos")
axes[0].legend()

axes[1].plot(k_values, relative_inertia_drop, marker="o", color="slateblue")
axes[1].axvline(k_elbow_sust, color="crimson", linestyle="--")
axes[1].set_xlabel("Número de clusters - k")
axes[1].set_ylabel("Caída relativa de inertia")
axes[1].set_title("Ganancia marginal al aumentar k")

plt.tight_layout()
plt.show()

print(f"Codo aproximado por heurística: k={k_elbow_sust}")
```

```python
K_coarse = range(10, 201, 10)
coarse_inertia = []

for k in K_coarse:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_sust)
    coarse_inertia.append(kmeans.inertia_)

coarse_k_values = np.array(list(K_coarse))
coarse_inertia_values = np.array(coarse_inertia)
coarse_relative_drop = np.r_[np.nan, (coarse_inertia_values[:-1] - coarse_inertia_values[1:]) / coarse_inertia_values[:-1]]

kmeans_sust_coarse_sweep = pd.DataFrame({
    "k": coarse_k_values,
    "inertia": coarse_inertia_values,
    "relative_inertia_drop": coarse_relative_drop,
})
kmeans_sust_coarse_sweep.to_csv(REPORTS_DIR / "kmeans_tfidf_sustantivos_k_coarse_sweep.csv", index=False)

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].plot(coarse_k_values, coarse_inertia_values, marker="o")
axes[0].set_xlabel("Número de clusters - k")
axes[0].set_ylabel("Inertia")
axes[0].set_title("Barrido amplio del codo: k=10..200")

axes[1].plot(coarse_k_values, coarse_relative_drop, marker="o", color="slateblue")
axes[1].set_xlabel("Número de clusters - k")
axes[1].set_ylabel("Caída relativa de inertia")
axes[1].set_title("Ganancia marginal cada 10 clusters")

plt.tight_layout()
plt.show()

kmeans_sust_coarse_sweep
```

La prueba amplia confirma que no aparece un codo nítido al llevar la exploración hasta `k=200`. La inercia sigue bajando, como es esperable al aumentar la cantidad de clusters, pero no se observa una estabilización clara ni un punto a partir del cual la ganancia marginal desaparezca. Por eso `k=12` no debe leerse como “el” número correcto de clusters: es apenas un candidato exploratorio de baja granularidad para poder mirar términos y ejemplos. A partir de este punto se usa `k=12` para todas las representaciones. No porque sea un óptimo claro, sino porque permite una comparación controlada: si cada representación usa un número distinto de clusters, la métrica mezcla dos cosas a la vez, la calidad de la representación y la granularidad elegida. Fijar `k` deja la pregunta más limpia: con la misma cantidad de grupos, ¿qué representación separa mejor y produce clusters más legibles?


```python
from sklearn.metrics import silhouette_score

silhouettes = []
cluster_size_summary = []

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_sust)
    score = silhouette_score(X_sust, labels)
    sizes = pd.Series(labels).value_counts()
    silhouettes.append(score)
    cluster_size_summary.append({
        "k": k,
        "min_cluster_size": int(sizes.min()),
        "median_cluster_size": float(sizes.median()),
        "max_cluster_size": int(sizes.max()),
    })

kmeans_sust_sweep = pd.DataFrame({
    "k": k_values,
    "inertia": inertia_values,
    "relative_inertia_drop": relative_inertia_drop,
    "silhouette": silhouettes,
    "elbow_distance": elbow_distances,
}).merge(pd.DataFrame(cluster_size_summary), on="k")

plt.figure(figsize=(8, 4))
plt.plot(k_values, silhouettes, marker="o", color="salmon")
plt.axvline(k_elbow_sust, color="crimson", linestyle="--", label=f"codo aprox.: k={k_elbow_sust}")
plt.title("Silhouette score")
plt.xlabel("Número de clusters (k)")
plt.ylabel("Silhouette")
plt.legend()
plt.show()

kmeans_sust_sweep.to_csv(REPORTS_DIR / "kmeans_tfidf_sustantivos_k_sweep.csv", index=False)

kmeans_sust_sweep.sort_values("silhouette", ascending=False).head(10)
```

Los términos representativos se calculan a nivel de cluster: se concatenan los textos de cada grupo y se aplica TF-IDF entre clusters. Esta lógica se parece a c-TF-IDF y ayuda a penalizar palabras que aparecen en casi todos los grupos.

La primera versión de esta tabla calculaba TF-IDF sobre todos los documentos individuales y después promediaba los pesos dentro de cada cluster. Eso era un problema para este análisis: como el corpus entero fue recolectado alrededor de obesidad y sobrepeso, palabras como `obesidad`, `obeso`, `obesa`, `sobrepeso`, `gordo` o `gordura` tenían peso alto en muchos tweets de muchos clusters. Al promediarlas por grupo, esos términos globales volvían a aparecer como “representativos” en casi todos los clusters, aunque no distinguieran realmente un tema de otro.

Por eso, para describir clusters no alcanza con preguntar qué palabras pesan mucho en los documentos del cluster; hay que preguntar qué palabras pesan más en ese cluster que en los demás. Al agregar los textos por cluster y calcular TF-IDF entre clusters, cada grupo compite contra los otros grupos como unidad temática. Así aparecen términos más diferenciales: por ejemplo referencias a salud, diabetes, política, comida, nacionalidad, deporte o insultos, en vez de repetir siempre las palabras que definieron la búsqueda inicial del dataset.

Además se excluyen términos demasiado generales del corpus, como `obesidad`, `obeso`, `obesa`, `sobrepeso`, `gordo` y variantes, para que la tabla muestre rasgos más distintivos de cada cluster.

```python
from sklearn.decomposition import TruncatedSVD
import matplotlib.pyplot as plt
import numpy as np

svd = TruncatedSVD(n_components=100, random_state=42)
svd.fit(X_sust)

explained_variance = svd.explained_variance_ratio_
cum_explained_variance = np.cumsum(explained_variance)

plt.figure(figsize=(8,5))
plt.plot(range(1, len(cum_explained_variance)+1), cum_explained_variance, marker='o')
plt.axhline(y=0.90, color='r', linestyle='--', label='90% varianza')
plt.axhline(y=0.95, color='g', linestyle='--', label='95% varianza')
plt.xlabel('Número de componentes SVD')
plt.ylabel('Varianza explicada acumulada')
plt.title('Selección de número de componentes SVD')
plt.legend()
plt.grid(True)
plt.show()
```

El objetivo de aplicar TruncatedSVD es reducir la dimensionalidad de la matriz TF-IDF, que suele ser muy dispersa y de alta dimensión en el caso de texto.

Para decidir el número de componentes a conservar, se graficó la varianza explicada acumulada. Este criterio busca encontrar el punto en el que agregar más dimensiones ya no aporta una ganancia significativa en términos de información retenida.

En este caso, el gráfico mostró que incluso con 100 componentes sólo se capta alrededor del 50% de la varianza. Esto es esperable en datos de texto, ya que la información está distribuida en muchísimas dimensiones.

A pesar de ello, se seleccionó un número intermedio de 80 componentes como compromiso entre mantener la mayor información posible de los datos, reducir la dimensionalidad para que los algoritmos sean computacionalmente más manejables, evitar el exceso de componentes que podrían reintroducir ruido y dificultar la interpretación.

En resumen, la decisión de usar 80 componentes fue pragmática, si bien no se alcanza un umbral clásico entre 90 y 95% de varianza explicada, este valor permite conservar suficiente información para explorar patrones sin perder eficiencia computacional.

Al evaluar KMeans con TF-IDF sobre sustantivos lematizados, la inercia disminuye a medida que aumenta el número de clusters, sin mostrar un punto de corte claro en `k=5`. En el rango inicial que se había probado (`k=2` a `k=19`), la curva seguía descendiendo; por eso no era correcto afirmar que `k=5` estuviera sostenido por el método del codo. Al ampliar la exploración hasta `k=60`, una heurística simple de distancia a la recta ubica un candidato cerca de `k=12`, pero el gráfico no muestra un codo fuerte.

La prueba adicional hasta `k=200`, saltando de a 10, refuerza esta lectura: la inercia continúa bajando y no aparece una estabilización evidente. El método del codo, en este caso, no ofrece una decisión clara; sirve más para descartar la idea de un corte obvio que para elegir un valor exacto.

La lectura principal es que TF-IDF sobre sustantivos sirve como línea base interpretable, pero es limitado para capturar la estructura temática de tweets cortos, ruidosos y con mucho contexto implícito. Esto motiva avanzar hacia representaciones más ricas, como Word2Vec y Sentence-BERT.

Word2Vec aprende vectores de palabras a partir de co-ocurrencias. En la variante skip-gram, la idea es usar una palabra para predecir las palabras de su contexto cercano:

$$\max \sum_{t=1}^{T} \sum_{-c \leq j \leq c, j \neq 0} \log p(w_{t+j}|w_t)$$

Luego cada tweet se resume promediando los vectores de sus palabras. Es una solución simple, pero ya incorpora relaciones semánticas que TF-IDF no puede capturar.

```python
def avg_cluster_similarity(embeddings, labels):
    """
    embeddings: array de vectores (tweets x dimensiones)
    labels: array con el cluster de cada tweet
    """
    clusters = np.unique(labels)
    avg_sims = {}

    for c in clusters:
        idx = np.where(labels == c)[0]
        cluster_emb = embeddings[idx]

        if len(cluster_emb) <= 1:
            avg_sims[c] = 1.0
            continue


        sim_matrix = cosine_similarity(cluster_emb)


        n = sim_matrix.shape[0]
        sum_sims = sim_matrix.sum() - n
        count = n * (n - 1)
        avg_sims[c] = sum_sims / count

    return avg_sims
```

```python
def avg_cluster_similarity(embeddings, labels):
    """
    embeddings: array de vectores (tweets x dimensiones)
    labels: array con el cluster de cada tweet
    """
    clusters = np.unique(labels)
    avg_sims = {}

    for c in clusters:
        idx = np.where(labels == c)[0]
        cluster_emb = embeddings[idx]

        if len(cluster_emb) <= 1:
            avg_sims[c] = 1.0  # solo un tweet, similaidad máxima
            continue


        sim_matrix = cosine_similarity(cluster_emb)


        n = sim_matrix.shape[0]
        sum_sims = sim_matrix.sum() - n
        count = n * (n - 1)
        avg_sims[c] = sum_sims / count

    return avg_sims
```

Al usar Sentence-BERT para generar embeddings de los tweets completos, se observa que los clusters capturan relaciones semánticas más complejas que TF-IDF. De todos modos, la separación visual sigue siendo difusa y la similitud interna no alcanza para hablar de grupos homogéneos. Esto es esperable en tweets cortos, irónicos y ruidosos: aunque el embedding contextual mejora la representación, KMeans sigue imponiendo particiones duras sobre un espacio donde los temas se mezclan.


Comparando las representaciones con el mismo `k=12`, la lectura deja de depender de elegir más o menos clusters para cada caso. TF-IDF aporta interpretabilidad, Word2Vec captura co-ocurrencias locales y Sentence-BERT incorpora contexto semántico. Aun así, ninguna variante muestra fronteras claras entre documentos; por eso esta sección se lee como exploración de representaciones y no como solución final de tópicos.
